## ❓ Troubleshooting Common Issues

````python
print("""
PROBLEM: "Out of memory" error in Google Colab
SOLUTION: 
  1. Reduce batch_size from 4 to 2
  2. Reduce max_seq_length from 2048 to 1024
  3. Use shorter training duration

PROBLEM: Model training is slow
SOLUTION:
  1. Make sure GPU is being used (check Colab settings)
  2. Reduce data size temporarily to test
  3. Use smaller model (if available)

PROBLEM: GGUF conversion fails
SOLUTION:
  1. Make sure LoRA weights are merged first
  2. Check file paths are correct
  3. Use Python 3.10+ for compatibility

PROBLEM: New model gives worse answers
SOLUTION:
  1. Training data quality issue (review pairs)
  2. Not enough training data (need 300+)
  3. Training time too short (run 3+ epochs)
  4. Try running for longer

PROBLEM: Model works but Streamlit crashes
SOLUTION:
  1. Make sure model path is correct
  2. Model file exists and is not corrupted
  3. Restart Streamlit: streamlit run streamlit_app.py --logger.level=debug
""")
````
