## 🎓 Key Concepts

### What is Fine-Tuning?
Taking a pre-trained model and teaching it specific domain knowledge. Like a piano player (model) learning new HR songs (data).

### Why Fine-Tune?
- Pre-trained models are general
- Your domain (HR policies) is specific
- Fine-tuning bridges the gap
- Results: better accuracy, fewer hallucinations

### Key Techniques

**LoRA (Low-Rank Adaptation)**
- Adds small trainable layers to frozen base model
- Saves 90% memory compared to full fine-tuning
- QLoRA = LoRA + Quantization (even more memory efficient)

**RAFT (Retrieval-Augmented FineTuning)**
- Teaches model to use retrieved documents better
- Reduces hallucination in RAG systems
- Newest and most RAG-optimized approach
